```python
import s3fs, h5py

ENDPOINT = "https://das.gein.noa.gr"
BUCKET = "neredis"
fs = s3fs.S3FileSystem(anon=True, client_kwargs={"endpoint_url": ENDPOINT})

print(fs.ls(BUCKET))                       # browse the bucket
```

```python
KEY = "asteri_2025-10-07T10_34_41+0100/asteri_2025-10-08T093541+0100.h5"                     # any HDF5 file in the bucket
CHANNEL, T0, T1 = 123, 0, 200              # channel index, sample window

with fs.open(f"{BUCKET}/{KEY}", "rb") as fh, h5py.File(fh) as f:
    ds = f["Acquisition/Raw[0]/RawData"]                         # (time, channel)
    trace = ds[T0:T1, CHANNEL]             # only these bytes are downloaded
    fs_hz = ds.attrs.get("fs")
print(trace.shape, fs_hz)
```

```python
import matplotlib.pyplot as plt
plt.plot(trace); plt.xlabel("sample"); plt.title(f"{KEY}  ch {CHANNEL}");
```
