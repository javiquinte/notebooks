```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import xarray as xr
import zarr
from obstore.store import S3Store
```

```python
store = S3Store(
    bucket="gc.3u2023zarr",
    endpoint="https://s3.gfz.de",
    skip_signature=True,
    virtual_hosted_style_request=False,
)
```

```python
s3store = zarr.storage.ObjectStore(store=store, read_only=True)
ds = xr.open_zarr(store=s3store, zarr_format=3, chunks=None)
```

```python
ds
```

```python
ds.strainrate[0:5, 5:10]
```

```python
ds.sel(channel="A0905")
```

```python
ds.sel(time=slice("2023-02-14T01:00:00", "2023-02-14T04:00:00"), channel="A0905")
```

```python
ds.isel(channel=(ds.lat > 52.3852) & (ds.lat < 52.3953))
```

```python
ds.isel(channel=(ds.distance > 3300) & (ds.distance < 3800)).sel(
    time=slice("2023-02-14T12:00:00", "2023-02-14T15:00:00")
)
```

```python
xr.plot.line(
    ds.sel(
        channel="A0905", time=slice("2023-02-14T12:00:00", "2023-02-14T15:00:00")
    ).strainrate,
    x="time",
    size=10,
)
```

```python
das2d = ds.isel(channel=slice(0, 930, 20)).sel(
    time=slice("2023-02-14T00:30:00", "2023-02-14T00:59:00")
)
```
