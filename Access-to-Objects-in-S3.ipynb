`boto3` is one of the libraries you could use from python to access data archived in some S3 Object Storage.

```python
import boto3
from botocore import UNSIGNED
from botocore.config import Config
```

It is very important to define who is actually hosting the data. Most of the examples you will find will be hosted in Amazon, but your Universities could also be providers of the S3 service.

```python
endpoint = "https://s3.gfz-potsdam.de/"
```

Pay attention to the `UNSIGNED` value in the configuration. In this particular case, you will access data that does not require to authenticate. This is not so usual in the S3 world. Usually you will be given a pair of keys (credentials) to operate with the service provider.

Within `boto3` there are 2 ways to operate on the buckets and objects. Through a `resource` or a `client`.

```python
s3r = boto3.resource(
    "s3", endpoint_url=endpoint, config=Config(signature_version=UNSIGNED)
)
```

Once the `resource` is created you can define the objects to operate with. For instace, a `Bucket`.

```python
bucket = s3r.Bucket("gc.3u2023")
```

The `Bucket` contains `objects` in it, that are the ones you would like to access to.

```python
for obj in bucket.objects.all():
    print("%4.1f MB" % (obj.size / (1024 * 1024),), obj.key)
```

Use the `download_file` method to get some object and save it physically in your local computer.

```python
bucket.download_file(Key="README.txt", Filename="README.txt")
```

Check what you got...

```python
fin = open("README.txt", "rt")
print(fin.read(100))
```

```python
fin.close()
```

The other way to operate was by defining a `Client`. Remember what we mentioned previously about specifying the location of the service provider and this special case of data available without credentials.

```python
s3c = boto3.client(
    "s3", endpoint_url=endpoint, config=Config(signature_version=UNSIGNED)
)
```

Use the `list_objects` method to see what is inside the bucket.

```python
objs = s3c.list_objects(Bucket="gc.3u2023")
```

Which information do you have available from each of these object? Check what is available for the first object.

```python
for k, v in objs["Contents"][0].items():
    print("%s: %s" % (k, v))
```

And finally display the key (name) of the objects in the bucket.

```python
for obj in objs["Contents"]:
    print(obj["Key"])
```

You can also use the `Client` with the `download_file` method.

```python
s3c.download_file("gc.3u2023", "3u2023.json", "3u2023.json")
```

And check what is inside the downloaded file.

```python
fin = open("3u2023.json", "rt")
print(fin.read(1000))
```
